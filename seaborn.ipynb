# Seaborn-Statical data Visualisation

## What is Seaborn?
Seaborn is a python library that buids on the top of matplotlib that makes the statistical data more easy and beautifull

## Why we need it?
1. For creating plots like voilin plot, box plot and pair plot
2. Autoomatically handles data from pandas data frames

## Where we use it?
1. Data Analysis and reporting
2. Exploring relationships between variables
3. Machine Learning for visualizing datasets and results


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=sns.load_dataset("titanic")
print(df.head())
print(df.tail())
print(df.columns)
print(df.shape)
print(df.info())
print(df.describe())
```

```python
## 4. Histograms & Countplots

# Histogram

sns.histplot(df['age'].dropna(), bins=10, kde=True)

plt.title("Histogram of Age")

plt.show()
```

```python

# Countplot

sns.countplot(x='class', data=df)

plt.title("Count of Passengers by Class")

plt.show()
```

```python
### 5. Scatter Plots

sns.scatterplot(x='age', y='fare', data=df, hue='sex')

plt.title("Age vs Fare")

plt.show()
```

```python
### 6. Boxplot & Violin Plot
# Boxplot
sns.boxplot(x='class', y='age', data=df)
plt.title("Boxplot of Age by Class")
plt.show()
```

```python
# Violin plot
sns.violinplot(x='class', y='age', hue='sex', data=df, split=True)
plt.title("Violin Plot of Age by Class and Sex")
plt.show()
```

```python
### 7. Pairplot
iris = sns.load_dataset('iris')
sns.pairplot(iris, hue='species')
plt.show()
```

```python
### 8. Heatmap
# Correlation heatmap
corr = iris.select_dtypes(include='number').corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
### 8. Heatmap
# Correlation heatmap (numeric columns only)
corr = iris.select_dtypes(include=['float64', 'int64']).corr()
 
import seaborn as sns
import matplotlib.pyplot as plt
 
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
### 9. Customization & Themes
sns.set_theme(style="whitegrid")  # options: darkgrid, white, ticks, etc.
sns.histplot(df['age'].dropna(), bins=10, color='orange')
plt.title("Customized Histogram with Seaborn")
plt.show()
 
```

```python
sns.barplot(x='class', y='fare', data=df, hue='sex', errorbar='sd')#instead of ci use errorbar
plt.title("Average Fare by Class and Sex")
plt.show()
```

```python
 
"Show trends or changes in mean values across categories."
sns.pointplot(x='class', y='fare', data=df, hue='sex', markers=['o', 'x'])
plt.title("Point Plot of Fare by Class and Sex")
plt.show()
 
```

```python
sns.kdeplot(df['fare'].dropna(), fill=True, color='teal')
plt.title("KDE Plot of Fare Distribution")
plt.show()
 
```

### 2. Import Libraries


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

### 3. Load Dataset

```python
df=sns.load_dataset("titanic")
df.head()
```

### 4. Explore Dataset

```python
print("Shape:", df.shape)
print(df.info())
print(df.isnull().sum())
print(df.describe())
```

### Insights:
 
### 1.There are 891 passengers with multiple features (age, sex, class, fare, etc.).
 
### 2.Missing values exist in age, deck, and embarked columns.
 
### 3.Understanding these helps guide data cleaning.

### 5. Data Clearing

```python
# Fill missing age with mean
df['age'] = df['age'].fillna(df['age'].mean())
# Fill embarked missing values with mode
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
# Drop rows with missing deck info
df = df.dropna(subset=['deck'])
```

### Insights:
 
1.Filling age ensures numeric analysis is accurate.
 
2.Filling embarked ensures categorical analysis works.
 
3.Dropping deck is optional since it has many missing values.

```python
# Average age of passengers
avg_age = np.mean(df['age'])
print("Average Age:", avg_age)
 
# Survival rate
survival_rate = df['survived'].mean()
print("Overall Survival Rate:", survival_rate)
 
# Survival rate by class
survival_by_class = df.groupby('class')['survived'].mean()
print("Survival Rate by Class:\n", survival_by_class)
 
# Survival rate by sex
survival_by_sex = df.groupby('sex')['survived'].mean()
print("Survival Rate by Sex:\n", survival_by_sex)
```

### Insights:
 
1.Average age: ~29 years.
 
2.Overall survival: ~38% passengers survived.
 
3.By class: 1st class had the highest survival, 3rd class the lowest.
 
4.By sex: Females had higher survival than males.

### 7. Visualization with Matplotlib
 

```python
# Bar chart of survival by class
classes = survival_by_class.index
rates = survival_by_class.values
 
plt.bar(classes, rates, color='skyblue')
plt.title("Survival Rate by Class")
plt.xlabel("Class")
plt.ylabel("Survival Rate")
plt.show()
```

#### Insights:
 
1.Visual confirms 2nd class passengers survived more.
 
2.Helps understand social/economic impact on survival.

```python
# Line plot of average fare by class
avg_fare = df.groupby('class')['fare'].mean()
plt.plot(avg_fare.index, avg_fare.values, marker='o', color='orange')
plt.title("Average Fare by Class")
plt.xlabel("Class")
plt.ylabel("Average Fare")
plt.show()
```

#### Insights:
 
1.1st class paid the highest fare, 3rd class the least.
 
2.Higher fare correlates with higher survival chances.

### 8. Visualization with Seaborn

```python
# Histogram of Age
sns.histplot(df['age'], bins=15, kde=True)
plt.title("Distribution of Age")
plt.show()
```

### Insights:
 
1.Most passengers were children/young adults, peak around 20-30 years

```python
# Countplot of survived by sex
sns.countplot(x='survived', hue='sex', data=df)
plt.title("Survival Count by Sex")
plt.show()
```

#### Insights:
 
1.Females survived more than males.
 
2.Confirms the "women and children first" pattern.

```python
# Boxplot of Fare by Class
sns.boxplot(x='class', y='fare', data=df)
plt.title("Fare Distribution by Class")
plt.show()
```

#### Insights:
 
1.1st class fares are more spread out, median higher.
 
2.3rd class has low fares with few outliers.

```python
# Pairplot of numeric features with survival hue
sns.pairplot(df[['age','fare','survived']], hue='survived')
plt.show()
```

### Insights:
 
1.Shows relationships between age, fare, and survival.
 
2.Higher fare seems associated with higher survival.

```python
# Heatmap of correlation
corr = df[['age','fare','survived']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
